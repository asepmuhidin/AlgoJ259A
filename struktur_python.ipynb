## Operator matematika

```python
#a=7
#b=3
a,b=7,3
#output 7 + 3 = 10
print(f"{a}+{b}={a+b}")
print(f"{a}-{b}={a-b}")
print(f"{a}x{b}={a*b}")
print(f"{a}/{b}={a/b}")
print(f"{a}//{b}={a//b}")
print(f"{a} mod {b}={a%b}")
print(f"{a} pangkat {b}={a**b}")
```
