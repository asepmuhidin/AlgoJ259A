```python
#Program menghitung Luas lingkaran

#INPUT
r=int(input("Masukkan nilai jari-jari lingkaran ?"))

#PROSES
luas=3.14*r*r

#OUTPUT
print(f"Luas lingkaran dengan r={r} adalah {luas}")
```

```python
#Program nilai lulus

#INPUT
nilai=int(input("Masukkan nilai anda ?"))

#PROSES
if nilai >=70:
  ket="lulus"
else:
  ket="Tidak lulus"

#OUTPUT
print(f"Nilai anda {nilai}, maka anda {ket}")


```

```python
#Program mencari nilai genap atau ganjil

#INPUT
N=int(input("Masukkan sembarang Nilai ?"))

#PROSES
if N%2==0:
  bilangan="Genap"
else :
  bilangan="Ganjil"

#OUTPUT
print(bilangan)
print(f"{N} adalah bilangan {bilangan}")
```
