```python
#a,b,c=1,-1,-6
a=int(input("Nilai a ? "))
b=int(input("Nilai b ? "))
c=int(input("Nilai c ? "))

d=b**2 - 4*a*c

print(f"Nilai D={d}")

# Jika (D > 0): memiliki 2 akar real berbeda.
# Jika (D = 0): memiliki 1 akar kembar.
# Jika (D < 0): akar-akarnya tidak real (imajiner)

if d>0:
  status="Memiliki 2 akar real berbeda"
  x1=(-b+math.sqrt(d))/(2*a)
  x2=(-b-math.sqrt(d))/(2*a)
elif d==0:
  status="Memiliki 1 akar kembar (x1=x2)"
  x1=(-b+math.sqrt(d))/(2*a)
  x2=x1
else:
  status="Akar-akrnya tidak real(imajiner)"
  x1='Imajiner'
  x2='Imajiner'

print(status)

print(f"Akar x1={x1}")
print(f"Akar x2={x2}")
```
