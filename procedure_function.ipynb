```python
import math

def akarKuadrat(a,b,c):

  d=b**2 - 4*a*c

  if d<0:
    return "x1 Imajiner","x1 Imajiner"

  else:
    x1=x1=(-b+math.sqrt(d))/(2*a)
    x2=(-b-math.sqrt(d))/(2*a)

    return x1,x2
```

```python
#a,b,c=1,-1,-6
a=int(input("Nilai a ? "))
b=int(input("Nilai b ? "))
c=int(input("Nilai c ? "))

x1,x2=akarKuadrat(a,b,c)
print()
print(f"Akar x1={x1}")
print(f"Akar x2={x2}")
```

#Nilai Akhir Angka dan hurup (NA/NH)
- NA=30%*uas+30%*uts+20%*tugas+20%*absen
- NH :
  - Jika NA >=85 , maka NH='A'
  - Jika NA >=75 , maka NH='B'
  - Jika NA >=60 , maka NH='C'
  - Jika NA >=45 , maka NH='D'
  - Jika NA < 45 , maka NH='E'



```python
def na(uas,uts,tugas,absen):
  return (0.3*uas)+(0.3*uts)+(0.2*tugas)+(0.2*absen)

def nh(nilai):
  grade=''

  if nilai>=85 :
    grade= 'A'
  elif nilai>=75 :
    grade= 'B'
  elif nilai>=60 :
    grade= 'C'
  elif nilai>=45 :
    grade= 'D'
  else:
    grade= 'E'

  return grade
```

```python
uas=int(input("Nilai UAS ? "))
uts=int(input("Nilai UTS ? "))
tugas=int(input("Nilai TUGAS ? "))
absen=int(input("Nilai ABSEN ? "))

nilai_angka=na(uas,uts,tugas,absen)

print()
print(f"Nilai anda :{nilai_angka}")

nilai_hurup=nh(nilai_angka)

print()
print(f"Grade : {nilai_hurup}")
```
