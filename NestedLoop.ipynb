```python
for i in range(3):
  for j in range(2):
    print(f"i={i}, j={j}")

```

```python
for i in range(1,6):
  for j in range(i):
    print("*", end="")
  print()
```

```python
for i in range(5,0,-1):
  for j in range(i):
    print("*", end="")
  print()
```

```python
for baris in range(1,6):
  for kolom in range(1,6):
    print(baris*kolom, end=" ")
  print()
```

```python
data=[
    [90,60,80,70],
    [50,40,90,100],
    [87,56,98,73],
]

for nilai in data:
  for nilai_mk in nilai:
    print(nilai_mk, end =" ")
  print()
```

```python
mat_A=[
    [4,5],
    [3,4]
]
mat_B=[
    [1,2],
    [2,3]
]

for baris in range(2):
  for kolom in range(2):
    print(mat_A[baris][kolom]+mat_B[baris][kolom], end= "")
  print()
```

```python
for a in mat_A:
  for b in mat_B:
    print(a+b, end= "")

```
